```python
# 1
# Write a function to check whether a number is even or odd
def check(a):
    if a % 2 == 0:
        print('even')
    else:
        print('odd')

check(4)
```

```python
# 2
# Write a function to print numbers from 1 to n
def print_numbers(n):
    for i in range(1, n + 1):
        print(i)

print_numbers(5)
```

```python
# 3
# Write a function to join a list of strings into a single string
def a(lst):
    result = ''
    for s in lst:
        result += s
    return result

print(a(['hello', ' ', 'world']))
```

```python
# 4
# Write a function to count vowels in a word
def b(x):
    v = 'AEIOUaeiou'
    count = 0
    for i in x:
        if i in v:
            count += 1
    print(count)

b('hello')
```

```python
# 5
# Write a function to reverse a given string
def a(x):
    return x[::-1]

print(a('python'))
```

```python
# 6
# Write a function to find maximum of three numbers
def a(x, y, z):
    if x >= y and x >= z:
        return x
    elif y >= x and y >= z:
        return y
    else:
        return z

print(a(5, 6, 7))
```

```python
# 7
# Write a function to multiply all numbers in a list
def sum_list(numbers):
    total = 1
    for n in numbers:
        total *= n
    return total

print(sum_list([1, 2, 3, 4, 5]))
```

```python
# 8
# Write a function that returns numbers greater than 10
def c(y):
    d = []
    for i in y:
        if i > 10:
            
            d.append(i)
    print(d)

c([5, 12, 3, 25, 10, 7, 18])
```

```python
# 9
# Write a function to check whether two words are anagrams
def a(x, y):
    return sorted(x) == sorted(y)

print(a('listen', 'silent'))
```

```python
# 10
# Write a function to check if a word is an isogram
def a(x):
    x = x.lower()
    return len(set(x)) == len(x)

print(a('Machine'))
print(a('Letter'))
```

```python
# 11
# Write a function to check if a word is a palindrome
def a(x):
    return x == x[::-1]

print(a('madam'))
print(a('hello'))
```

```python
# 12
# Count vowels and consonants ignoring case and spaces
def a(x):
    x = x.lower()
    v = 0
    c = 0
    for i in x:
        if i == ' ':
            continue
        elif i in 'aeiou':
            v += 1
        else:
            c += 1
    print('vowels:', v)
    print('consonants:', c)

a('Hello World')
```

```python
# 13
# Check whether a string is an isogram
def a(x):
    x = x.lower()
    for i in x:
        if x.count(i) > 1:
            return False
    return True

print(a('Machine'))
print(a('Letter'))
```

```python
# 14
# Return the longest word in a sentence
def a(x):
    words = x.split()
    longest = words[0]
    for w in words:
        if len(w) > len(longest):
            longest = w
    return longest

print(a('I love Python programming'))
```

```python
# 15
# Remove all punctuation from a string
def a(x):
    p = '.,!?;:\'"()-'
    result = ''
    for i in x:
        if i not in p:
            result += i
    return result

print(a('Hello, World! How are you?'))
```

```python
# 16
# Return string where every character is repeated twice
def a(x):
    result = ''
    for i in x:
        result += i * 2
    return result

print(a('cat'))
```

```python
# 17
# Find missing number from 1 to n
def a(arr, n):
    total = n * (n + 1) // 2
    return total - sum(arr)

print(a([1, 2, 4, 5], 5))
```

```python
# 18
# Return second largest number without sorting
def a(lst):
    largest = -1
    second = -1
    for i in lst:
        if i > largest:
            second = largest
            largest = i
        elif i > second and i != largest:
            second = i
    return second

print(a([5, 2, 8, 1, 9]))
```

```python
# 19
# Rotate list to the left by k positions
def a(arr, k):
    k = k % len(arr)
    return arr[k:] + arr[:k]

print(a([1, 2, 3, 4, 5], 2))
```

```python
# 20
# Move all zeros to end maintaining order
def a(lst):
    d = []
    z = 0
    for i in lst:
        if i != 0:
            d.append(i)
        else:
            z += 1
    d += [0] * z
    print(d)

a([0, 1, 0, 3, 12])
```

```python
# 21
# Return common elements from two lists without sets
def a(x, y):
    d = []
    for i in x:
        if i in y and i not in d:
            d.append(i)
    return d

print(a([1, 2, 3, 4], [2, 4, 6]))
```

```python
# 22
# Recursive function to compute nth Fibonacci number
def a(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return a(n - 1) + a(n - 2)

print(a(7))
```

```python
# 23
# Recursive function to find sum of digits
def a(n):
    if n == 0:
        return 0
    return n % 10 + a(n // 10)

print(a(1234))
```

```python
# 24
# Return list of prime factors
def a(n):
    d = []
    i = 2
    while i <= n:
        while n % i == 0:
            d.append(i)
            n = n // i
        i += 1
    return d

print(a(36))
```

```python
# 25
# Merge two dictionaries summing common key values
def a(d1, d2):
    result = dict(d1)
    for key in d2:
        if key in result:
            result[key] += d2[key]
        else:
            result[key] = d2[key]
    return result

print(a({'a': 1, 'b': 2}, {'b': 3, 'c': 4}))
```

```python
# 26
# Return first non-repeating character
def a(x):
    for i in x:
        if x.count(i) == 1:
            return i
    return None

print(a('aabbcde'))
```

```python
# 27
# Invert dictionary swapping keys and values
def a(d):
    result = {}
    for key in d:
        val = d[key]
        if val in result:
            result[val].append(key)
        else:
            result[val] = [key]
    return result

print(a({'a': 1, 'b': 2, 'c': 1}))
```

```python
# 28
# Return students who scored above average
def a(d):
    avg = sum(d.values()) / len(d)
    result = []
    for name in d:
        if d[name] > avg:
            result.append(name)
    return result

scores = {'Akshay': 85, 'Rahul': 60, 'Priya': 90, 'Sneha': 70}
print(a(scores))
```

```python
# 29
# Group words by their starting character
def a(words):
    result = {}
    for word in words:
        key = word[0]
        if key in result:
            result[key].append(word)
        else:
            result[key] = [word]
    return result

print(a(['apple', 'banana', 'avocado', 'blueberry', 'cherry']))
```
