```python
#5 Write a function to calculate the sum of squares of numbers in a list.
def a(lst):
    total = 0
    for i in lst:
        total += i**2
    return total
a([1,2,3,4,5])    
```

```python
#6 whether the string is Symmetrical and a palindrome at the same time.
def a(x):
  return x == x [::-1]
a('malayalam')
```
