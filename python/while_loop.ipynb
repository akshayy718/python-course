```python
# 1
# Find the factorial of a number
a = int(input('enter a number'))
factorial = 1
i = 1
while i <= a:
    factorial *= i
    i += 1
print(factorial)
```

```python
# 2
# Generate the Fibonacci series up to N terms
n = int(input('enter number of terms'))
a = 0
b = 1
count = 0
while count < n:
    print(a)
    c = a + b
    a = b
    b = c
    count += 1
```

```python
# 3
# Check if a number is Armstrong using while loop
a = int(input('enter a number'))
original = a
total = 0
digits = len(str(a))
while a > 0:
    digit = a % 10
    total += digit ** digits
    a = a // 10
if total == original:
    print('armstrong number')
else:
    print('not armstrong number')
```

```python
# 4
# Check if a number is prime using while loop
a = int(input('enter a number'))
i = 2
is_prime = True
while i < a:
    if a % i == 0:
        is_prime = False
        break
    i += 1
if is_prime and a > 1:
    print('prime number')
else:
    print('not prime number')
```

```python
# 6
# Find the GCD of two numbers using while loop
a = int(input('enter first number'))
b = int(input('enter second number'))
while b != 0:
    a, b = b, a % b
print(a)
```

```python
# 7
# Find the LCM of two numbers using while loop
a = int(input('enter first number'))
b = int(input('enter second number'))
lcm = max(a, b)
while True:
    if lcm % a == 0 and lcm % b == 0:
        print(lcm)
        break
    lcm += 1
```

```python
# 8
# Convert a decimal number to binary
a = int(input('enter a decimal number'))
b = ''
while a > 0:
    b = str(a % 2) + b
    a = a // 2
print(b)
```

```python
# 9
# Convert a binary number to decimal
a = input('enter a binary number')
decimal = 0
power = 0
i = len(a) - 1
while i >= 0:
    decimal += int(a[i]) * (2 ** power)
    power += 1
    i -= 1
print(decimal)
```

```python
# 10
# Find the frequency of a digit in a number
a = int(input('enter a number'))
d = int(input('enter digit to find'))
count = 0
while a > 0:
    if a % 10 == d:
        count += 1
    a = a // 10
print(count)
```

```python
# 11
# Remove all zeros from a number
a = int(input('enter a number'))
b = ''
while a > 0:
    digit = a % 10
    if digit != 0:
        b = str(digit) + b
    a = a // 10
print(int(b))
```

```python
# 12
# Count even and odd digits in a number
a = int(input('enter a number'))
even = 0
odd = 0
while a > 0:
    digit = a % 10
    if digit % 2 == 0:
        even += 1
    else:
        odd += 1
    a = a // 10
print('even digits:', even)
print('odd digits:', odd)
```

```python
# 13
# Find the second largest digit in a number
a = int(input('enter a number'))
largest = -1
second = -1
while a > 0:
    digit = a % 10
    if digit > largest:
        second = largest
        largest = digit
    elif digit > second and digit != largest:
        second = digit
    a = a // 10
print('second largest digit:', second)
```

```python
# 14
# Calculate a^b without using built-in functions
a = int(input('enter base'))
b = int(input('enter power'))
result = 1
i = 0
while i < b:
    result *= a
    i += 1
print(result)
```

```python
# 15
# Keep asking for numbers until 0, then print sum
sum = 0
while True:
    a = int(input('enter a number (0 to stop)'))
    if a == 0:
        break
    sum += a
print('sum:', sum)
```

```python
# 18
# Number pattern increasing using while loop
i = 1
while i <= 5:
    j = 1
    while j <= i:
        print(j, end='')
        j += 1
    print()
    i += 1
```

```python
# 19
# Repeated number pattern using while loop
i = 1
while i <= 5:
    print(str(i) * i)
    i += 1
```

```python
# 20
# Floyd's Triangle using while loop
n = 1
i = 1
while i <= 5:
    j = 1
    while j <= i:
        print(n, end=' ')
        n += 1
        j += 1
    print()
    i += 1
```

```python
# 21
# Number pattern decreasing using while loop
i = 5
while i >= 1:
    j = 1
    while j <= i:
        print(j, end='')
        j += 1
    print()
    i -= 1
```

```python
# 22
# Pyramid pattern using while loop
rows = 5
i = 1
while i <= rows:
    print(' ' * (rows - i) + '*' * (2 * i - 1))
    i += 1
```

```python
# 23
# Inverted pyramid using while loop
rows = 5
i = rows
while i >= 1:
    print(' ' * (rows - i) + '*' * (2 * i - 1))
    i -= 1
```

```python
# 24
# Hollow square using while loop
n = 5
i = 1
while i <= n:
    if i == 1 or i == n:
        print('*' * n)
    else:
        print('*' + ' ' * (n - 2) + '*')
    i += 1
```

```python
# 27
# Alphabet pattern decreasing using while loop
i = 5
while i >= 1:
    j = 0
    while j < i:
        print(chr(65 + j), end='')
        j += 1
    print()
    i -= 1
```

```python
# 28
# Repeated number with spaces using while loop
i = 1
while i <= 5:
    j = 0
    while j < i:
        print(i, end=' ')
        j += 1
    print()
    i += 1
```

```python
# 29
# Number pattern decreasing from 5 using while loop
i = 1
while i <= 5:
    j = 5
    while j >= 5 - i + 1:
        print(j, end='')
        j -= 1
    print()
    i += 1
```

```python
# 30
# Diamond number pattern using while loop
rows = 5
i = 1
while i <= rows:
    print(' ' * (rows - i) + ' '.join(str(j) for j in range(1, i + 1)) + ' '.join(str(j) for j in range(i - 1, 0, -1)))
    i += 1
```

```python
# 31
# Hollow rectangle using while loop
rows = 4
cols = 6
i = 1
while i <= rows:
    if i == 1 or i == rows:
        print('*' * cols)
    else:
        print('*' + ' ' * (cols - 2) + '*')
    i += 1
```
