```python
# 17
print("1. Check Even/Odd")
print("2. Find Largest of Two Numbers")
print("3. Check Leap Year")
print("4. Exit")

choice = int(input("Enter your choice: "))

if choice == 1:
    num = int(input("Enter a number: "))
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")

elif choice == 2:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    if a > b:
        print("Largest:", a)
    else:
        print("Largest:", b)

elif choice == 3:
    year = int(input("Enter year: "))
    if year % 4 == 0:
        print("Leap year")
    else:
        print("Not a leap year")

elif choice == 4:
    print("Exiting...")

else:
    print("Invalid choice")
```

```python
# 18
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in meters: "))
bmi = weight / (height ** 2)
print("Your BMI:", round(bmi, 2))
if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

```python
# 19
amount = float(input("Enter purchase amount: "))
if amount >= 5000:
    discount = amount * 0.20
    print("20% discount! You save:", discount)
    print("Final amount:", amount - discount)
elif amount >= 2000:
    discount = amount * 0.10
    print("10% discount! You save:", discount)
    print("Final amount:", amount - discount)
else:
    print("No discount. Final amount:", amount)
```
