```python
# 12
# Check whether a number is an Armstrong number

a = int(input('enter a number'))
total = 0
digits = len(str(a))

for i in str(a):
    total += int(i) ** digits

if total == a:
    print('armstrong number')
else:
    print('not armstrong number')
```

```python
# 13
# Check whether a number is a palindrome

a = input('enter a number')
b = ''

for i in a:
    b = i + b

if a == b:
    print('palindrome')
else:
    print('not palindrome')
```

```python
# 17
# Find the sum of all even numbers in a given range

a = int(input('enter start'))
b = int(input('enter end'))
sum = 0

for i in range(a, b + 1):
    if i % 2 == 0:
        sum += i

print(sum)
```

```python
# 18
# Print all leap years between two given years

a = int(input('enter start year'))
b = int(input('enter end year'))

for i in range(a, b + 1):
    if (i % 4 == 0 and i % 100 != 0) or i % 400 == 0:
        print(i)
```

```python
# 19
# Find the average of n numbers entered by the user

n = int(input('how many numbers'))
sum = 0

for i in range(n):
    a = int(input('enter number'))
    sum += a

avg = sum / n

print(avg)
```

```python
# 20
# Check whether a number is prime

a = int(input('enter a number'))
is_prime = True

for i in range(2, a):
    if a % i == 0:
        is_prime = False
        break

if is_prime and a > 1:
    print('prime number')
else:
    print('not prime number')
```

```python
# 21
# Print all prime numbers between 1 and 100

for a in range(2, 101):
    is_prime = True

    for i in range(2, a):
        if a % i == 0:
            is_prime = False
            break

    if is_prime:
        print(a)
```

```python
# 22
# Find the GCD of two numbers using a loop

a = int(input('enter first number'))
b = int(input('enter second number'))

smaller = min(a, b)
gcd = 1

for i in range(1, smaller + 1):
    if a % i == 0 and b % i == 0:
        gcd = i

print(gcd)
```

```python
# 23
# Find the LCM of two numbers using a loop

a = int(input('enter first number'))
b = int(input('enter second number'))

for i in range(max(a, b), a * b + 1):
    if i % a == 0 and i % b == 0:
        lcm = i
        break

print(lcm)
```

```python
# 26
# Print number pattern increasing
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```
