```python
# Reading a file

file = open("demo.txt", "r")
print(file.read())
file.close()
```

```python
# Appending to the file

file = open("demo.txt", "a")
file.write("\nthis is a demoooooooooo")
file.close()

file = open("demo.txt", "r")
print(file.read())
file.close()
```

```python
# Writing to the file

file = open("demo.txt", "w")
file.write("this is a new sent")
file.close()

file = open("demo.txt", "r")
print(file.read())
file.close()
```
