```python
# 1
# Student Management System using list, dict, tuple, functions, for loop, while loop

students = []

def add_student():
    name = input('enter name')
    age = int(input('enter age'))
    marks = tuple(int(input(f'enter mark {i+1}')) for i in range(3))
    student = {'name': name, 'age': age, 'marks': marks}
    students.append(student)
    print('student added!')

def display_students():
    if len(students) == 0:
        print('no students found')
    else:
        for s in students:
            print('name:', s['name'])
            print('age:', s['age'])
            print('marks:', s['marks'])
            print('---')

def find_topper():
    if len(students) == 0:
        print('no students found')
    else:
        topper = students[0]
        for s in students:
            if sum(s['marks']) > sum(topper['marks']):
                topper = s
        print('topper:', topper['name'])
        print('marks:', topper['marks'])

def avg_marks():
    if len(students) == 0:
        print('no students found')
    else:
        for s in students:
            avg = sum(s['marks']) / len(s['marks'])
            print(s['name'], 'average:', avg)

while True:
    print('1. add student')
    print('2. display students')
    print('3. find topper')
    print('4. average marks')
    print('5. exit')
    choice = int(input('enter choice'))
    if choice == 1:
        add_student()
    elif choice == 2:
        display_students()
    elif choice == 3:
        find_topper()
    elif choice == 4:
        avg_marks()
    elif choice == 5:
        print('goodbye!')
        break
    else:
        print('invalid choice')
```

```python
# 2
# ATM Machine using functions, if-else, while loop

balance = 10000
pin = 1234

def check_balance():
    print('your balance is:', balance)

def deposit():
    global balance
    amount = int(input('enter amount to deposit'))
    if amount > 0:
        balance += amount
        print('deposited successfully!')
        print('new balance:', balance)
    else:
        print('invalid amount')

def withdraw():
    global balance
    amount = int(input('enter amount to withdraw'))
    if amount <= 0:
        print('invalid amount')
    elif amount > balance:
        print('insufficient balance')
    else:
        balance -= amount
        print('withdrawn successfully!')
        print('remaining balance:', balance)

def change_pin():
    global pin
    old = int(input('enter old pin'))
    if old == pin:
        new = int(input('enter new pin'))
        pin = new
        print('pin changed successfully!')
    else:
        print('incorrect pin')

while True:
    entered_pin = int(input('enter your pin'))
    if entered_pin != pin:
        print('wrong pin!')
        continue
    print('1. check balance')
    print('2. deposit money')
    print('3. withdraw money')
    print('4. change pin')
    print('5. exit')
    choice = int(input('enter choice'))
    if choice == 1:
        check_balance()
    elif choice == 2:
        deposit()
    elif choice == 3:
        withdraw()
    elif choice == 4:
        change_pin()
    elif choice == 5:
        print('thank you!')
        break
    else:
        print('invalid choice')
```

```python
# 3
# Calculator using functions, if-else, while loop

def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    if b == 0:
        print('cannot divide by zero')
    else:
        return a / b

while True:
    print('1. addition')
    print('2. subtraction')
    print('3. multiplication')
    print('4. division')
    print('5. exit')
    choice = int(input('enter choice'))
    if choice == 5:
        print('goodbye!')
        break
    elif choice in [1, 2, 3, 4]:
        a = float(input('enter first number'))
        b = float(input('enter second number'))
        if choice == 1:
            print('result:', add(a, b))
        elif choice == 2:
            print('result:', sub(a, b))
        elif choice == 3:
            print('result:', mul(a, b))
        elif choice == 4:
            print('result:', div(a, b))
    else:
        print('invalid choice')
```

```python
# OOP 1
# ATM class with private pin

class ATM:
    def __init__(self):
        self.__pin = 1234

    def verify_pin(self, pin):
        if pin == self.__pin:
            print('pin verified!')
            return True
        else:
            print('wrong pin!')
            return False

    def change_pin(self, old_pin, new_pin):
        if old_pin == self.__pin:
            self.__pin = new_pin
            print('pin changed successfully!')
        else:
            print('incorrect old pin')

a = ATM()
a.verify_pin(1234)
a.change_pin(1234, 5678)
a.verify_pin(5678)
```

```python
# OOP 2
# Student class with private marks

class Student:
    def __init__(self):
        self.__marks = 0

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
            print('marks updated!')
        else:
            print('invalid marks! must be between 0 and 100')

    def get_marks(self):
        print('marks:', self.__marks)

s = Student()
s.set_marks(85)
s.get_marks()
s.set_marks(150)
```

```python
# OOP 3
# Login class with public username and private password

class Login:
    def __init__(self, username, password):
        self.username = username
        self.__password = password

    def verify_password(self, password):
        if password == self.__password:
            print('login successful!')
            return True
        else:
            print('wrong password!')
            return False

    def change_password(self, old_password, new_password):
        if old_password == self.__password:
            self.__password = new_password
            print('password changed successfully!')
        else:
            print('incorrect old password')

l = Login('akshay', 'pass123')
l.verify_password('pass123')
l.change_password('pass123', 'newpass456')
l.verify_password('newpass456')
```

```python
# OOP 4
# StudentPortal class with public name and private cgpa

class StudentPortal:
    def __init__(self, student_name):
        self.student_name = student_name
        self.__cgpa = 0.0

    def update_cgpa(self, cgpa):
        if 0 <= cgpa <= 10:
            self.__cgpa = cgpa
            print('cgpa updated!')
        else:
            print('invalid cgpa! must be between 0 and 10')

    def view_profile(self):
        print('name:', self.student_name)
        print('cgpa:', self.__cgpa)

s = StudentPortal('Akshay')
s.update_cgpa(8.5)
s.view_profile()
s.update_cgpa(11)
```

```python
# OOP 5
# Online Shopping Website with private cart total

class OnlineShoppingWebsite:
    def __init__(self):
        self.__cart_total = 0

    def add_to_cart(self, amount):
        if amount > 0:
            self.__cart_total += amount
            print('added to cart! amount:', amount)
        else:
            print('invalid amount')

    def get_total(self):
        print('cart total:', self.__cart_total)

    def checkout(self):
        if self.__cart_total == 0:
            print('your cart is empty!')
        else:
            print('order placed! total amount:', self.__cart_total)
            self.__cart_total = 0
            print('cart cleared!')

s = OnlineShoppingWebsite()
s.add_to_cart(500)
s.add_to_cart(1200)
s.get_total()
s.checkout()
s.get_total()
```

```python
# Inheritance 4
# Grandparent, Parent, Child - multi level inheritance

class Grandparent:
    def show_grandparent(self):
        print('i am grandparent')

class Parent(Grandparent):
    def show_parent(self):
        print('i am parent')

class Child(Parent):
    def show_child(self):
        print('i am child')

c = Child()
c.show_grandparent()
c.show_parent()
c.show_child()
```

```python
# Inheritance 5
# Animal, Mammal, Dog - multi level inheritance

class Animal:
    def eat(self):
        print('eating')

class Mammal(Animal):
    def walk(self):
        print('walking')

class Dog(Mammal):
    def bark(self):
        print('barking')

d = Dog()
d.eat()
d.walk()
d.bark()
```

```python
# Polymorphism 1
# Animal sounds - Cow and Sheep

class Animal:
    def speak(self):
        print('Animal speaks')

class Cow(Animal):
    def speak(self):
        print('Cow says Moo')

class Sheep(Animal):
    def speak(self):
        print('Sheep says Baa')

for animal in [Cow(), Sheep(), Animal()]:
    animal.speak()
```

```python
# Polymorphism 2
# Vehicle horn - Car and Bike

class Vehicle:
    def horn(self):
        print('Vehicle horn')

class Car(Vehicle):
    def horn(self):
        print('Car horn: Beep Beep')

class Bike(Vehicle):
    def horn(self):
        print('Bike horn: Peep Peep')

for v in [Car(), Bike()]:
    v.horn()
```

```python
# Polymorphism 3
# Shape perimeter - Square and Triangle

class Shape:
    def perimeter(self):
        pass

class Square(Shape):
    def perimeter(self):
        side = 5
        print('Square perimeter:', 4 * side)

class Triangle(Shape):
    def perimeter(self):
        s1, s2, s3 = 3, 4, 5
        print('Triangle perimeter:', s1 + s2 + s3)

for s in [Square(), Triangle()]:
    s.perimeter()
```

```python
# Polymorphism 4
# Shape area - Rectangle and Circle

class Shape:
    def area(self):
        pass

class Rectangle(Shape):
    def area(self):
        length = float(input('enter length'))
        width = float(input('enter width'))
        print('rectangle area:', length * width)

class Circle(Shape):
    def area(self):
        radius = float(input('enter radius'))
        print('circle area:', 3.14 * radius ** 2)

for s in [Rectangle(), Circle()]:
    s.area()
```
