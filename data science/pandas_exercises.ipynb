```python
import pandas as pd
df = pd.read_csv(r'C:\Users\aksha\OneDrive\Documents\python_exercises\python\students_20.csv')
print(df)
#We used head() to get a quick overview of the dataset — to check column names, data types, and sample values before doing any analysis.
```

```python
df[['Student Name', 'Math']]
#We selected only these two columns to focus on the specific information needed, avoiding unnecessary data in the output.
```

```python
print('Math average:', df['Math'].mean())
print('Science average:', df['Science'].mean())
print('English average:', df['English'].mean())
#We calculated the mean of each subject to understand the overall performance level of the class in each subject.
```

```python
highest_science = df[df['Science'] == df['Science'].max()]
print(highest_science[['Student Name', 'Science']])
#We used max() to find the highest Science score and then filtered the DataFrame to identify which student achieved it.
```

```python
result = df[df['English'] > 90]
print(result[['Student Name', 'English']])
#We applied a filter condition to extract only the high-performing students in English, which helps identify top achievers.
```

```python
sorted_df = df.sort_values('Math', ascending=False)
print(sorted_df[['Student Name', 'Math']])
# We sorted the data in descending order to rank students from highest to lowest Math score, making it easy to identify the top performers.
```

```python
# Q7 - Add Total column = Math + Science + English
# We created a Total column to get a single value representing each student's overall performance
df['Total'] = df['Math'] + df['Science'] + df['English']
print(df[['Student Name', 'Math', 'Science', 'English', 'Total']])
```

```python
# Q7 - Add Total column = Math + Science + English
# We created a Total column to get a single value representing each student's overall performance
df['Total'] = df['Math'] + df['Science'] + df['English']
print(df[['Student Name', 'Math', 'Science', 'English', 'Total']])
```

```python
# Q8 - Student with highest total marks
# We used max() on Total column and filtered to find the best overall performing student
highest = df[df['Total'] == df['Total'].max()]
print(highest[['Student Name', 'Total']])
```

```python
# Q9 - Students who scored less than 80 in at least one subject
# We used OR condition to identify students who need improvement in at least one subject
result = df[(df['Math'] < 80) | (df['Science'] < 80) | (df['English'] < 80)]
print(result[['Student Name', 'Math', 'Science', 'English']])
```

```python
# Q10 - Save updated DataFrame to new CSV
# We saved the updated DataFrame with Total column so the data can be reused without losing changes
df.to_csv('students_with_totals.csv', index=False)
print('Saved successfully!')
```

```python
# Q11 - Highest and lowest mark in Math
# We used max() and min() to find the range of Math scores to understand variation in student performance
print('Highest Math:', df['Math'].max())
print('Lowest Math:', df['Math'].min())
```

```python
# Q12 - Average total marks of all students
# We calculated the average total to get a benchmark of what a typical student scores overall
print('Average Total:', df['Total'].mean())
```

```python
# Q13 - Top 3 students in English
# We sorted by English marks and used head(3) to extract only top 3 to recognise highest achievers
top3 = df.sort_values('English', ascending=False).head(3)
print(top3[['Student Name', 'English']])
```

```python
# Q14 - Count students who scored above 85 in Science
# We filtered and counted to understand what proportion of class is performing well in Science
count = len(df[df['Science'] > 85])
print('Students above 85 in Science:', count)
```

```python
# Q15 - Create Grade column based on Total marks
# We classified each student into grades using a custom function and apply() for easy comparison
def get_grade(total):
    if total >= 270:
        return 'A'
    elif total >= 240:
        return 'B'
    elif total >= 210:
        return 'C'
    else:
        return 'D'

df['Grade'] = df['Total'].apply(get_grade)
print(df[['Student Name', 'Total', 'Grade']])
```

```python
# Q16 - Average marks per subject for each Grade
# We used groupby() to group students by Grade and find which grade performs better in which subject
result = df.groupby('Grade')[['Math', 'Science', 'English']].mean()
print(result)
```

```python
# Q17 - Students who have same marks in any two subjects
# We compared each pair of subject columns to find unusual patterns in student scores
result = df[
    (df['Math'] == df['Science']) |
    (df['Science'] == df['English']) |
    (df['Math'] == df['English'])
]
print(result[['Student Name', 'Math', 'Science', 'English']])
```

```python
# Q18 - Top 5 students based on total marks
# We sorted by Total and used head(5) to get a clear ranking of the best overall performers
top5 = df.sort_values('Total', ascending=False).head(5)
print(top5[['Student Name', 'Total']])
```

```python
# Q19 - Save Student Name and Grade to new CSV
# We selected only two columns and saved as a separate CSV to create a clean grade report
df[['Student Name', 'Grade']].to_csv('students_grades.csv', index=False)
print('Grades CSV saved successfully!')
```
