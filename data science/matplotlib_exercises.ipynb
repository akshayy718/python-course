```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create student data instead of company sales
data = {
    'month': [1, 2, 3, 4, 5, 6],
    'maths': [85, 90, 78, 92, 88, 95],
    'science': [70, 75, 80, 85, 72, 88],
    'english': [60, 65, 70, 75, 80, 85],
    'history': [55, 60, 65, 70, 75, 80],
    'total': [270, 290, 293, 322, 315, 348]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Basic data analysis
print(df.shape)         # rows and columns
print(df.describe())    # statistical summary
print(df.head())        # first 5 rows
```

```python
# Line chart - monthly marks for each subject
plt.figure(figsize=(10, 6))

plt.plot(df['month'], df['maths'], label='Maths', marker='o')
plt.plot(df['month'], df['science'], label='Science', marker='o')
plt.plot(df['month'], df['english'], label='English', marker='o')
plt.plot(df['month'], df['history'], label='History', marker='o')

plt.title('Monthly Subject Marks')
plt.xlabel('Month')
plt.ylabel('Marks')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Bar chart - total marks per subject
subject_totals = df[['maths', 'science', 'english', 'history']].sum()

subject_totals.plot(kind='bar', figsize=(8, 5), color='coral')
plt.title('Total Marks per Subject')
plt.ylabel('Total Marks')
plt.xlabel('Subject')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Pie chart - subject wise marks distribution
plt.figure(figsize=(6, 6))

plt.pie(
    subject_totals,
    labels=subject_totals.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title('Subject-wise Marks Distribution')
plt.show()
```

```python
# Histogram of total marks
plt.figure(figsize=(8, 6))

plt.hist(df['total'], bins=5, color='purple', edgecolor='black')

plt.xlabel('Total Marks', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Histogram of Total Marks', fontsize=14)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
